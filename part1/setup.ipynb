## Hugging Face access

Accept the conditions on the [WildGuardMix dataset page](https://huggingface.co/datasets/allenai/wildguardmix) with your Hugging Face account. The real-data notebook uses that account's existing token.

The following cell checks for a token without printing it. If you need to log in, run `hf auth login` in the terminal with this environment active. Alternatively, set `log_in_now = True` below; it asks for the token through a hidden input. Never put a token directly in notebook code or output.

```python
from huggingface_hub import get_token, login
from getpass import getpass

log_in_now = False
if log_in_now:
    login(token=getpass("Hugging Face token (hidden): "), add_to_git_credential=False)
print("A local Hugging Face token is configured:", get_token() is not None)
```
